```python
import pandas as pd
```

```python
callsign = []
position = []
velocity = []

with open("../output.txt", "r", encoding="utf-8") as file:
    for line in file:
        clean_line = line.rstrip("\n")

        msg = int(clean_line, 16)
        typecode = (msg >> 75) & 0x1F

        if 1 <= typecode <= 4:
            callsign.append((clean_line, typecode))

        elif 5 <= typecode <= 18:
            position.append((clean_line, typecode))

        elif typecode == 19:
            velocity.append((clean_line, typecode))
```

```python
print(len(callsign))
print(len(position))
print(len(velocity))
```
